**REQUIRED INITIAL CODES:**

```python
import pandas as pd #This imports the Panda Library and uses the "pd" keyword to shorten the codings.
```

```python
df = pd.read_csv('board2.csv') #Reads the board2.csv file and creates a DataFrame named df
df #Displays the new DataFrame
```

```python
Average = df[["Math", "Electronics", "GEAS", "Communication"]].mean(axis=1) #Gets the mean of the required column labels.
Average #Shows the calculated average.
```

```python
df = df.assign(Average=Average) #Adds the calculated Average as a new column in the DataFrame
df #Displays the read DataFrame
```

# **A. VISAYAS COMMUNICATION DATAFRAME**

Create a DataFrame named VisComm containing students whose Hometown is Visayas and whose Track
is Communication.

```python
VisComm = df[(df["Hometown"] == "Visayas") & (df["Track"] == "Communication")][["Name", "Gender", "Math", "Electronics", "Average"]] #Only shows the students whose Hometown is Visayas AND whose Track is Communication, then selects only required colummns.
VisComm #Displays the VisComm DataFrame.
```

# **B. VISAYAS FEMALE DATAFRAME**

Create a second DataFrame named VisFemale containing students whose Hometown is Visayas and whose Gender is Female whose Average is at least 60.

```python
VisFemale = df[(df["Hometown"] == "Visayas") & (df["Gender"] == "Female")][["Name", "Track", "GEAS", "Electronics", "Average"]] #Only shows the students whose Hometown is Visayas AND whose Track is Communication, then selects only required colummns.
display(VisFemale) #Displays the VisComm DataFrame.
```

```python
VisFemale_60 = VisFemale[VisFemale["Average"] >= 60] #Filters VisFemale to show only Students from the VisFemale DataFrame of at least
display(VisFemale_60) #Displays the Visayas Female DataFrame
```

# **C. CATEGORY-AVERAGE VISUALIZATION**

***C.1. For each feature, compute the mean of Average for every category using Pandas.***

```python
Communication = df[df['Track'] == 'Communication']['Average'].mean() #Calculates the mean Average of Communication students
print('Communication:', Communication) #Displays the Communication mean Average
Instrumentation = df[df['Track'] == 'Instrumentation']['Average'].mean() #Calculates the mean Average of Instrumentation students
print('Instrumentation:', Instrumentation) #Displays the Instrumentation mean Average
Microelectronics = df[df['Track'] == 'Microelectronics']['Average'].mean() #Calculates the mean Average of Microelectronics students
print('Microelectronics :', Microelectronics) #Displays the Microelectronics mean Average
```

```python
Female = df[df['Gender'] == 'Female']['Average'].mean() #Calculates the mean Average of Female students
print('Female:', Female) #Displays the Female mean Average
Male = df[df['Gender'] == 'Male']['Average'].mean() #Calculates the mean Average of Male students
print('Male:', Male) #Displays the Male mean Average
```

```python
Luzon = df[df['Hometown'] == 'Luzon']['Average'].mean() #Calculates the mean Average of students from Luzon
print('Luzon:', Luzon) #Displays the Luzon mean Average
Visayas = df[df['Hometown'] == 'Visayas']['Average'].mean() #Calculates the mean Average of students from Visayas
print('Visayas:', Visayas) #Displays the Visayas mean Average
Mindanao = df[df['Hometown'] == 'Mindanao']['Average'].mean() #Calculates the mean Average of students from Mindanao
print('Mindanao:', Mindanao) #Displays the Mindanao mean Average
```

***C.2. Display the three summary tables.***

```python
#Creates new DataFrame containing the mean Average for each Track
TrackMean = pd.DataFrame({
    'Track': ['Communication', 'Instrumentation', 'Microelectronics'],
    'Mean Average': [Communication, Instrumentation, Microelectronics]
})
TrackMean #Displays the new TrackMean DataFrame
```

```python
#Creates a DataFrame containing the mean Average for each Gender
GenderMean = pd.DataFrame({
    'Gender': ['Female', 'Male'],
    'Mean Average': [Female, Male]
})
GenderMean #Displays the new GenderMean DataFrame
```

```python
#Creates a DataFrame containing the mean Average for each Hometown
HometownMean = pd.DataFrame({
    'Hometown': ['Luzon', 'Visayas', 'Mindanao'],
    'Mean Average': [Luzon, Visayas, Mindanao]
})
HometownMean #Displays the new HometownMean DataFrame
```

***C.3. Create one figure containing three bar charts: mean Average by Track, by Gender, and by
Hometown.***

```python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 3, figsize=(15, 5)) #Creates one figure containing three bar charts with size of 15 x 5


# Track
TrackMean.plot(kind='bar', ax=axes[0]) #Creates a bar chart showing the mean Average for each Track
axes[0].set_title('Mean Average by Track') #Sets the title of the Track graph
axes[0].set_xlabel('Track') #Sets the x-axis label
axes[0].set_ylabel('Mean Average') #Sets the y-axis label

# Gender
GenderMean.plot(kind='bar', ax=axes[1]) #Creates a bar chart showing the mean Average for each Gender
axes[1].set_title('Mean Average by Gender') #Sets the title of the Gender graph
axes[1].set_xlabel('Gender') #Sets the x-axis label
axes[1].set_ylabel('Mean Average') #Sets the y-axis label

# Hometown
HometownMean.plot(kind='bar', ax=axes[2]) #Creates a bar chart showing the mean Average for each Hometown
axes[2].set_title('Mean Average by Hometown') #Sets the title of the Hometown graph
axes[2].set_xlabel('Hometown') #Sets the x-axis label
axes[2].set_ylabel('Mean Average') #Sets the y-axis label

plt.tight_layout() #Adjusts the spacing between the three graphs
plt.show() #Displays the completed figure
```

```python
#Finds the category with the highest sample mean Average, Gender and Hometown
HighestTrackMean = TrackMean.loc[TrackMean['Mean Average'].idxmax(), 'Track']
HighestGenderMean = GenderMean.loc[GenderMean['Mean Average'].idxmax(), 'Gender']
HighestHometownMean = HometownMean.loc[HometownMean['Mean Average'].idxmax(), 'Hometown']

#Displays the category with the highest sample mean
print(HighestTrackMean, "has the highest sample mean Average garnering 67.975 among the Track categories.")
print(HighestGenderMean, "has the highest sample mean Average amounting to 65.225 among the Gender categories.")
print(HighestHometownMean, "has the highest sample mean Average of 67.500 among the Hometown categories.")
```
